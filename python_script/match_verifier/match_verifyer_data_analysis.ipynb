```python
scored = df.dropna(subset=["similarity_score"])

# Step 2: Select only those with similarity_score < 0.8
low_score_hashes = scored.loc[scored["similarity_score"] < 0.8, "hash"]

# Step 3: Extract *both* rows for those hashes (the pair)
pairs = df[df["hash"].isin(low_score_hashes)]
```

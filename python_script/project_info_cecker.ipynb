```python
#### Method Checker ####

import json
import os
import requests
from datetime import datetime
from dotenv import load_dotenv

load_dotenv("../src/config/.env")

TOKEN = os.getenv("JSON_TOKEN")
ADDRESS = os.getenv("JSON_URL")

#content =  {"ProjectInfo": [{"project_id":1891391718,"version":1483196148000}]}
# 0077377f-cca5-4479-1842-3bd03932fc31
# 6467b3e013df3d68492c59e4

def to_human_time(ts):
    if ts > 1e12:  # assume milliseconds
        ts = ts / 1000
    return datetime.fromtimestamp(ts).strftime('%Y-%m-%d %H:%M:%S')


def checkerFun(hashVal):
    content = { "Check": [hashVal]}


    data = {
        "token": TOKEN,
        "content": content
    }

    headers = {
        'Content-Type': 'application/json'
    }


    try:
        #print("in try..")
        response = requests.post(ADDRESS, json=data, headers=headers)


        if response.status_code == 200:
            result_list = response.json()
            print("Success:", result_list)
             # Process the first item if it's a list of results
            if isinstance(result_list, list) and len(result_list) > 0:
                first_result = result_list[0]

                sv_time = first_result.get('sv_time')
                ev_time = first_result.get('ev_time')

                print(sv_time, ev_time)

                if sv_time is not None and ev_time is not None:
                    sv_time_human = to_human_time(sv_time)
                    ev_time_human = to_human_time(ev_time)
                    print(f"Start Time: {sv_time_human}")
                    print(f"End Time:   {ev_time_human}")
                else:
                    print("Start or end time not found in first result.")
        else:
            print(f"Failed with status code {response.status_code}")
            print("Response:", response.text)

    except requests.exceptions.RequestException as e:
        print(f"Error during request: {e}")
```
